```python
def years_code_status(years_code):

    try:
        years = int(years_code)

        if years < 3:
            return "Beginner"

        if years < 6:
            return "Early Career"

        if years < 11:
            return "Experienced"

        if years >= 11:
            return "Highly Experienced"
        return "Unknown"

    except:
        return "Unknown"
```

```python
df["experienceLevel"] = df["yearsCode"].apply(years_code_status)
```

```python
df["experienceLevel"].explode().unique()
```

```python
def usesDocumentation_bool(methods_list, target_option):
    if not isinstance(methods_list, list):
                return False
    return target_option in methods_list

```

```python
DOC_OPTION = (
        "Technical documentation (is generated for/by the tool or system)"
    )
AI_OPTION = "AI CodeGen tools or AI-enabled apps"
SO_OPTION = "Stack Overflow or Stack Exchange"
```

```python

df["usesDocumentation"] = df["learningMethods"].apply(
        lambda x: usesDocumentation_bool(x, DOC_OPTION)
    )
df["usesAIForLearning"] = df["learningMethods"].apply(
        lambda x: usesDocumentation_bool(x, AI_OPTION)
    )
df["usesStackOverflow"] = df["learningMethods"].apply(
        lambda x: usesDocumentation_bool(x, SO_OPTION)
    )
```
