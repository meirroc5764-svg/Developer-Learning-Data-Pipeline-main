```python
def years_code_status(years_code):

    try:
        years = int(years_code)

        if years < 3:
            return "Beginner"

        if years < 6:
            return "Early Career"

        if years < 11:
            return "Experienced"

        if years > 11:
            return "Highly Experienced"
        return "Unknown"

    except:
        return "Unknown"
```

```python
df["experienceLevel"] = df["yearsCode"].apply(years_code_status)
```
